 Q1. Create the first numpy array of 1-D with values (1,2,3,0). 
     Create a second numpy array of 1-D with values (4,1,0, 6).
     Perform following simple arithmetic operations on these two numpy arrays: 
     • Addition 
     • Subtraction 
     • Multiplication 
     • Division 

```python
import numpy as np

# Create the first 1-D array
arr1 = np.array([1, 2, 3, 0])

# Create the second 1-D array
arr2 = np.array([4, 1, 0, 6])

# Perform arithmetic operations
print("Addition:", arr1 + arr2)
print("Subtraction:", arr1 - arr2)
print("Multiplication:", arr1 * arr2)

# Safe division (avoid division by zero errors)
division = np.divide(arr1, arr2, out=np.zeros_like(arr1, dtype=float), where=arr2!=0)
print("Division:", division)
```

Q2. Create the first numpy array of random integers between 1-100 of 5 elements in 1-D.
    Create second numpy array of 2-D for the same range of integers of shape 4*5.
    Now perform following simple arithmetic operations on these two numpy arrays: 
    • Addition 
    • Subtraction 

```python
import numpy as np

# 1-D array of 5 random integers between 1 and 100
arr1 = np.random.randint(1, 101, size=5)

# 2-D array of shape 4x5 with random integers between 1 and 100
arr2 = np.random.randint(1, 101, size=(4, 5))

# Perform arithmetic operations
addition = arr2 + arr1   # Broadcasting: arr1 is added to each row of arr2
subtraction = arr2 - arr1  # Broadcasting: arr1 is subtracted from each row of arr2

# Print results
print("1-D Array:\n", arr1)
print("\n2-D Array:\n", arr2)
print("\nAddition:\n", addition)
print("\nSubtraction:\n", subtraction)
```

Q3. Create a 3-D array of 27 random elements using random function of shape 3*3*3.
    Now pull the following elements using indexing/slicing 
    • Third column of second 2-D array 
    • First and third row of first 2-D array 
    • Intersection of first row and third column in third 2-D aray 

```python
import numpy as np

# Create a 3-D array of shape (3,3,3) with random integers between 1 and 100
arr3d = np.random.randint(1, 101, size=(3, 3, 3))

print("3-D Array:\n", arr3d)

# 1. Third column of second 2-D array
third_col_second_2d = arr3d[1, :, 2]

# 2. First and third row of first 2-D array
first_third_rows_first_2d = arr3d[0, [0, 2], :]

# 3. Intersection of first row and third column in third 2-D array
intersection = arr3d[2, 0, 2]

print("\nThird column of second 2-D array:\n", third_col_second_2d)
print("\nFirst and third row of first 2-D array:\n", first_third_rows_first_2d)
print("\nIntersection (first row, third column, third 2-D array):\n", intersection)
```

Q4. Create a 2-D numpy array of random integers between 1 to 100 with shape 5*4.
    Now perform following operations: 
    • Print only the even numbers from this array 
    • Print only the odd numbers from this array 
    • Print row wise sum of all elements 
    • Print column wise sum of all elements 
    • Convert this 2-D array into 3-D array of shape 2*2*5 

```python
import numpy as np

# Create a 2-D array of shape (5,4) with random integers between 1 and 100
arr2d = np.random.randint(1, 101, size=(5, 4))

print("Original 2-D Array:\n", arr2d)

# 1. Print only the even numbers
even_numbers = arr2d[arr2d % 2 == 0]
print("\nEven Numbers:\n", even_numbers)

# 2. Print only the odd numbers
odd_numbers = arr2d[arr2d % 2 != 0]
print("\nOdd Numbers:\n", odd_numbers)

# 3. Row-wise sum of all elements
row_sum = np.sum(arr2d, axis=1)
print("\nRow-wise Sum:\n", row_sum)

# 4. Column-wise sum of all elements
col_sum = np.sum(arr2d, axis=0)
print("\nColumn-wise Sum:\n", col_sum)

# 5. Convert this 2-D array into 3-D array of shape (2,2,5)
arr3d = arr2d.reshape(2, 2, 5)
print("\nConverted 3-D Array of shape (2,2,5):\n", arr3d)
```

Q5. Create a 2-D numpy array of random integers between 1 to 100 with shape 5*4.
    Now perform following operations: 
    • Swap first row with third row 
    • Swap second column with fourth column. 
    • Replace the values less than 50 with zero. 
    • Convert this 2D array into 1-D array 

```python
import numpy as np

# Create a 2-D array of shape (5,4) with random integers between 1 and 100
arr2d = np.random.randint(1, 101, size=(5, 4))

print("Original 2-D Array:\n", arr2d)

# 1. Swap first row with third row
arr2d[[0, 2]] = arr2d[[2, 0]]
print("\nAfter Swapping First and Third Row:\n", arr2d)

# 2. Swap second column with fourth column
arr2d[:, [1, 3]] = arr2d[:, [3, 1]]
print("\nAfter Swapping Second and Fourth Column:\n", arr2d)

# 3. Replace values less than 50 with zero
arr2d[arr2d < 50] = 0
print("\nAfter Replacing Values < 50 with 0:\n", arr2d)

# 4. Convert this 2-D array into 1-D array
arr1d = arr2d.flatten()
print("\nConverted 1-D Array:\n", arr1d)
```

Q6. Explain the concept of stacking in numpy with examples. 
    How is it different from concatenation. Give any 3 differences between them. 

• Stacking means combining arrays along a new axis.
• It increases the dimensionality of the result.
• Common functions:
      np.stack() → general stacking along a new axis
      np.vstack() → vertical stacking (row-wise)
      np.hstack() → horizontal stacking (column-wise)
      np.dstack() → depth stacking (along the third axis)


```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Stack along a new axis (axis=0)
stacked0 = np.stack((a, b), axis=0)
print("Stacked along axis=0:\n", stacked0)

# Stack along a new axis (axis=1)
stacked1 = np.stack((a, b), axis=1)
print("\nStacked along axis=1:\n", stacked1)
```

```python
# Vertical stacking (rows)
vstacked = np.vstack((a, b))
print("\nVertical Stack:\n", vstacked)

# Horizontal stacking (columns)
hstacked = np.hstack((a, b))
print("\nHorizontal Stack:\n", hstacked)
```
