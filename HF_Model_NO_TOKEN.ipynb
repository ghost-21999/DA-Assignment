```python
SENTIMENT_MODEL = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"
QA_MODEL = "distilbert-base-cased-distilled-squad"
GENERATION_MODEL = "distilgpt2"
MASK_MODEL = "distilroberta-base"

print("Model names set successfully!")
```

```python
print("Loading Hugging Face pipelines...")

sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model=SENTIMENT_MODEL
)

qa_pipeline = pipeline(
    "question-answering",
    model=QA_MODEL
)

generation_pipeline = pipeline(
    "text-generation",
    model=GENERATION_MODEL
)

mask_pipeline = pipeline(
    "fill-mask",
    model=MASK_MODEL
)

print("All pipelines loaded successfully!")
```

```python
print("Loading AutoTokenizer and AutoModel...")

sentiment_tokenizer = AutoTokenizer.from_pretrained(
    SENTIMENT_MODEL
)

sentiment_model = AutoModelForSequenceClassification.from_pretrained(
    SENTIMENT_MODEL
)

print("AutoTokenizer and AutoModel loaded successfully!")
```

```python
def sentiment_with_automodel(text):
    start_time = time.time()

    inputs = sentiment_tokenizer(
        text,
        return_tensors="pt",
        truncation=True
    )

    with torch.no_grad():
        outputs = sentiment_model(**inputs)

    probabilities = torch.softmax(outputs.logits, dim=-1)

    predicted_class = torch.argmax(probabilities, dim=-1).item()

    labels = sentiment_model.config.id2label
    label = labels[predicted_class]

    confidence = probabilities[0][predicted_class].item()

    inference_time = time.time() - start_time

    return label, confidence, inference_time
```

```python
def compare_sentiment(text):
    if not text or not text.strip():
        return "Please enter some text."

    # Pipeline implementation
    start_pipeline = time.time()

    pipeline_result = sentiment_pipeline(text)[0]

    pipeline_time = time.time() - start_pipeline

    # AutoModel implementation
    automodel_label, automodel_confidence, automodel_time = (
        sentiment_with_automodel(text)
    )

    result = f"""
PIPELINE IMPLEMENTATION
-----------------------
Model: {SENTIMENT_MODEL}
Prediction: {pipeline_result['label']}
Confidence: {pipeline_result['score']:.4f}
Inference Time: {pipeline_time:.4f} seconds


AUTOMODEL IMPLEMENTATION
------------------------
Model: {SENTIMENT_MODEL}
Prediction: {automodel_label}
Confidence: {automodel_confidence:.4f}
Inference Time: {automodel_time:.4f} seconds


COMPARISON
----------
Pipeline Prediction: {pipeline_result['label']}
AutoModel Prediction: {automodel_label}

Pipeline Time: {pipeline_time:.4f} seconds
AutoModel Time: {automodel_time:.4f} seconds
"""

    return result
```

```python
def question_answering(context, question):
    if not context.strip():
        return "Please enter a context."

    if not question.strip():
        return "Please enter a question."

    try:
        start_time = time.time()

        result = qa_pipeline(
            question=question,
            context=context
        )

        inference_time = time.time() - start_time

        return f"""
Model: {QA_MODEL}

Answer: {result['answer']}

Confidence Score: {result['score']:.4f}

Inference Time: {inference_time:.4f} seconds
"""

    except Exception as e:
        return f"Error: {str(e)}"
```

```python
def text_generation(prompt):
    if not prompt.strip():
        return "Please enter a prompt."

    try:
        start_time = time.time()

        result = generation_pipeline(
            prompt,
            max_new_tokens=50,
            num_return_sequences=1
        )

        inference_time = time.time() - start_time

        generated_text = result[0]["generated_text"]

        return f"""
Model: {GENERATION_MODEL}

Generated Text:
{generated_text}

Inference Time: {inference_time:.4f} seconds
"""

    except Exception as e:
        return f"Error: {str(e)}"
```

```python
def mask_word_prediction(text):
    if not text.strip():
        return "Please enter a sentence."

    if "<mask>" not in text:
        return "Please include <mask> in your sentence."

    try:
        start_time = time.time()

        results = mask_pipeline(
            text,
            top_k=5
        )

        inference_time = time.time() - start_time

        output = f"""
Model: {MASK_MODEL}

Top Predictions:
"""

        for i, result in enumerate(results, start=1):
            output += (
                f"{i}. {result['token_str'].strip()} "
                f"— Confidence: {result['score']:.4f}\n"
            )

        output += f"\nInference Time: {inference_time:.4f} seconds"

        return output

    except Exception as e:
        return f"Error: {str(e)}"
```

```python
def text_intelligence_assistant(
    task,
    text,
    context="",
    question=""
):

    try:

        if task == "Sentiment Analysis":
            return compare_sentiment(text)

        elif task == "Question Answering":
            return question_answering(
                context,
                question
            )

        elif task == "Text Generation":
            return text_generation(text)

        elif task == "Mask Word Prediction":
            return mask_word_prediction(text)

        else:
            return "Invalid task selected."

    except Exception as e:
        return f"Unexpected error: {str(e)}"
```

```python
PROMPT_TEMPLATES = {

    "Zero-Shot": """
Answer the following user query directly and accurately.

User Query:
{query}

Provide a clear, useful and well-structured answer.
""",

    "One-Shot": """
You are given one example showing the desired answering style.

Example:
User: What is Artificial Intelligence?
Assistant: Artificial Intelligence is the field of computer science concerned with creating systems that can perform tasks that normally require human intelligence, such as learning, reasoning and decision-making.

Now answer the following query in a similar clear and informative style.

User Query:
{query}
""",

    "Few-Shot": """
Use the following examples as demonstrations of the desired response style.

Example 1:
Question: What is machine learning?
Answer: Machine learning is a branch of AI in which systems learn patterns from data and use those patterns to make predictions or decisions.

Example 2:
Question: What is deep learning?
Answer: Deep learning is a type of machine learning that uses multi-layer neural networks to learn complex patterns from large datasets.

Example 3:
Question: What is NLP?
Answer: Natural Language Processing is a field of AI that enables computers to understand, process and generate human language.

Now answer this query using the same concise, structured style.

User Query:
{query}
""",

    "Chain-of-Thought": """
Solve the following problem carefully.

User Query:
{query}

Analyze the problem systematically before giving the answer.
Provide the key reasoning steps or justification, but do not reveal private chain-of-thought.
Give the final answer clearly and concisely.
""",

    "Role-Based / Persona": """
You are an expert subject-matter consultant and an excellent teacher.

Your responsibilities:
- Explain concepts accurately.
- Use appropriate examples.
- Organize information clearly.
- Adapt the explanation to the user's question.
- Avoid unnecessary complexity.

User Query:
{query}

Answer as the expert persona described above.
"""
}

print("Prompt templates created:", len(PROMPT_TEMPLATES))
```

```python
def generate_response(prompt, temperature, max_tokens, top_p, top_k):
    """Generate text locally with the Transformers pipeline; no API token required."""
    try:
        local_generator = generation_pipeline
        generated = local_generator(
            prompt,
            max_new_tokens=min(int(max_tokens), 200),
            do_sample=True,
            temperature=max(0.1, float(temperature)),
            top_p=float(top_p),
            top_k=int(top_k),
            pad_token_id=local_generator.tokenizer.eos_token_id
        )
        full_text = generated[0]["generated_text"]
        answer = full_text[len(prompt):].strip() if full_text.startswith(prompt) else full_text.strip()
        return answer if answer else "The local model returned an empty response. Try a shorter prompt."
    except Exception as error:
        return (
            "Local generation could not run: " + str(error) +
            "\\nMake sure Assignment 1.1 model-loading cells completed successfully."
        )
```

```python
def evaluate_response(query, response, strategy):

    if not response or response.startswith("ERROR"):
        return {
            "Accuracy": 0,
            "Completeness": 0,
            "Structure": 0,
            "Creativity": 0,
            "Reasoning Quality": 0,
            "Overall": 0
        }

    text = response.strip()

    # Completeness
    completeness = min(100, 40 + min(len(text) / 8, 40))

    # Structure
    structure_points = 0

    if len(text.splitlines()) > 2:
        structure_points += 25

    if re.search(r"\n[-*•]", text):
        structure_points += 25

    if re.search(r"\d+\.", text):
        structure_points += 25

    if ":" in text:
        structure_points += 25

    structure = min(100, structure_points)

    # Creativity
    creativity = 60

    creative_words = [
        "example",
        "analogy",
        "scenario",
        "imagine",
        "for instance"
    ]

    creativity += sum(
        8 for word in creative_words
        if word in text.lower()
    )

    creativity = min(100, creativity)

    # Reasoning quality
    reasoning = 55

    reasoning_words = [
        "because",
        "therefore",
        "however",
        "step",
        "reason",
        "first",
        "second",
        "finally"
    ]

    reasoning += sum(
        7 for word in reasoning_words
        if word in text.lower()
    )

    reasoning = min(100, reasoning)

    # Basic accuracy proxy
    accuracy = 75

    if len(text) > 100:
        accuracy += 10

    if len(text.split()) > 40:
        accuracy += 10

    accuracy = min(100, accuracy)

    overall = round(
        (
            accuracy +
            completeness +
            structure +
            creativity +
            reasoning
        ) / 5,
        2
    )

    return {
        "Accuracy": round(accuracy, 2),
        "Completeness": round(completeness, 2),
        "Structure": round(structure, 2),
        "Creativity": round(creativity, 2),
        "Reasoning Quality": round(reasoning, 2),
        "Overall": overall
    }
```

```python
def run_strategy(query, strategy, temperature, max_tokens, top_p, top_k):

    if not query.strip():
        return "", "Please enter a query."

    template = PROMPT_TEMPLATES[strategy]

    prompt = template.format(query=query)

    response = generate_response(
        prompt,
        temperature,
        max_tokens,
        top_p,
        top_k
    )

    scores = evaluate_response(
        query,
        response,
        strategy
    )

    score_text = "\n".join([
        f"Accuracy: {scores['Accuracy']}",
        f"Completeness: {scores['Completeness']}",
        f"Structure: {scores['Structure']}",
        f"Creativity: {scores['Creativity']}",
        f"Reasoning Quality: {scores['Reasoning Quality']}",
        f"Overall: {scores['Overall']}"
    ])

    return response, score_text
```

```python
def compare_strategies(
    query,
    temperature,
    max_tokens,
    top_p,
    top_k
):

    if not query.strip():
        return (
            "",
            "",
            "",
            "",
            "",
            pd.DataFrame()
        )

    outputs = {}
    rows = []

    for strategy, template in PROMPT_TEMPLATES.items():

        prompt = template.format(query=query)

        response = generate_response(
            prompt,
            temperature,
            max_tokens,
            top_p,
            top_k
        )

        scores = evaluate_response(
            query,
            response,
            strategy
        )

        outputs[strategy] = {
            "prompt": prompt,
            "response": response
        }

        rows.append({
            "Strategy": strategy,
            "Accuracy": scores["Accuracy"],
            "Completeness": scores["Completeness"],
            "Structure": scores["Structure"],
            "Creativity": scores["Creativity"],
            "Reasoning Quality": scores["Reasoning Quality"],
            "Overall": scores["Overall"]
        })

    return (
        outputs["Zero-Shot"]["prompt"],
        outputs["Zero-Shot"]["response"],
        outputs["One-Shot"]["response"],
        outputs["Few-Shot"]["response"],
        outputs["Chain-of-Thought"]["response"],
        pd.DataFrame(rows)
    )
```

```python
def optimize_prompt(query, strategy):

    template = PROMPT_TEMPLATES[strategy]

    optimized = f"""
You are an expert prompt engineer.

Task:
Answer the user's query accurately and clearly.

User Query:
{query}

Instructions:
1. Understand the intent of the query.
2. Provide a direct answer.
3. Organize the response with headings or bullet points when useful.
4. Include relevant examples where appropriate.
5. Avoid unsupported claims.
6. Keep the answer focused and useful.

Prompting Strategy:
{strategy}

Generate the best possible response.
"""

    return optimized.strip()
```

```python
prompt_history = []


def save_history(query, strategy, prompt, response, scores):

    prompt_history.append({
        "Time": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "Query": query,
        "Strategy": strategy,
        "Prompt": prompt,
        "Response": response,
        "Overall Score": scores["Overall"]
    })

    return pd.DataFrame(prompt_history)
```

```python
def generate_selected(
    query,
    strategy,
    temperature,
    max_tokens,
    top_p,
    top_k
):

    prompt = PROMPT_TEMPLATES[strategy].format(
        query=query
    )

    response = generate_response(
        prompt,
        temperature,
        max_tokens,
        top_p,
        top_k
    )

    scores = evaluate_response(
        query,
        response,
        strategy
    )

    score_text = "\n".join([
        f"Accuracy: {scores['Accuracy']}",
        f"Completeness: {scores['Completeness']}",
        f"Structure: {scores['Structure']}",
        f"Creativity: {scores['Creativity']}",
        f"Reasoning Quality: {scores['Reasoning Quality']}",
        f"Overall: {scores['Overall']}"
    ])

    history = save_history(
        query,
        strategy,
        prompt,
        response,
        scores
    )

    return (
        prompt,
        response,
        score_text,
        history
    )
```
