```python
#g)	Delete all the rows where Director column has missing values.

print(df.columns.tolist())
df.columns = df.columns.str.strip()
print("Missing values in Director column:", df['Director'].isnull().sum())
df = df.dropna(subset=['Director'])
print("Missing values in Director column after cleaning:", df['Director'].isnull().sum())
print("New shape of dataframe:", df.shape)
```

```python
#h)	Print all the records where country has Germany value (including West Germany). 
#If any other country is there along with Germany, then that row should also come in output.

germany_records = df[df['country'].str.contains("Germany", case=False, na=False)]
print(germany_records)
print(germany_records['country'])
print("Number of records with Germany:", germany_records.shape[0])
```

```python
print(df['Duration'].unique())
```

```python
#i)	Expand Duration column into 2 separate columns – First column having the numeric value and other having String. 
#Eg: 3 seasons should be split in 2 columns having 3 in 1st column and seasons in 2nd column.

df.columns = df.columns.str.strip()

unique_durations = df['Duration'].unique()

durations_df = pd.DataFrame(unique_durations, columns=['Duration'])

durations_df['Duration_Value'] = durations_df['Duration'].str.extract(r'(\d+)')
durations_df['Duration_Unit'] = durations_df['Duration'].str.extract(r'([A-Za-z]+)')

durations_df['Duration_Value'] = pd.to_numeric(durations_df['Duration_Value'], errors='coerce')

print(durations_df)
```

```python
#j)	Split Date added into 3 separate columns having date value in 1st column, month value in 2nd column and year value in 3rd.

# Normalize column names (strip spaces, lowercase)
df.columns = df.columns.str.strip().str.lower()

# Convert 'date_added' to datetime safely
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

# Expand into Day, Month, Year
df['Day'] = df['date_added'].dt.day
df['Month'] = df['date_added'].dt.month_name()   # full month name
df['Year'] = df['date_added'].dt.year

print(df[['date_added', 'Day', 'Month', 'Year']])
```

```python
#k)	Print the number of TV shows/Movies released in each year.

release_counts = df.groupby(['release_year', 'type']).size().reset_index(name='count')

print(release_counts)
```

```python
#l)	Rename the column title with movie_title.

df = df.rename(columns={'title': 'movie_title'})

print(df.columns.tolist())
```

```python
#m)	Split Listed_in column into 3 different columns with col name (Genre1, Genre2, Genre3). Split the column based on comma.

df.columns = df.columns.str.strip().str.lower()

df[['Genre1', 'Genre2', 'Genre3']] = df['listed_in'].str.split(',', n=2, expand=True)

df['Genre1'] = df['Genre1'].str.strip()
df['Genre2'] = df['Genre2'].str.strip()
df['Genre3'] = df['Genre3'].str.strip()

print(df[['listed_in', 'Genre1', 'Genre2', 'Genre3']])
```

# Ques.2 Import both the attached files (student.csv and marks.csv) in Jupyter notebook and perform following operations:

```python
#a)	Combine both the dataframes into single dataframe which will have all the records from both the tables
# Load both CSV files
marks_df = pd.read_csv("marks.csv")
student_df = pd.read_csv("student.csv")

# Merge on Student_id
combined_df = pd.merge(marks_df, student_df, on="Student_id", how="outer")

# Display first 10 rows
print(combined_df.head(10))
```

```python
# Load both CSV files
marks_df = pd.read_csv("marks.csv")
student_df = pd.read_csv("student.csv")

# Merge on Student_id
combined_df = pd.merge(marks_df, student_df, on="Student_id", how="outer")

# Save to a new CSV file
combined_df.to_csv("combined.csv", index=False)

print("Combined dataframe saved as combined.csv")
```

```python
#b)	Print the maximum and minimum marks Gender wise.

df = pd.read_csv("combined.csv")

# Group by Gender and calculate max & min marks
result = df.groupby("Gender")["Mark"].agg(["max", "min"])

print(result)
```

```python
#c)	Print all the students IDs and their marks who have scored more than the average marks of the class

# Load the combined file
df = pd.read_csv("combined.csv")

# Calculate average marks (ignoring NaN)
avg_marks = df["Mark"].mean()
print("Class Average Marks:", avg_marks)

# Filter students scoring above average
above_avg = df[df["Mark"] > avg_marks][["Student_id", "Mark"]]

print(above_avg)
```

```python
#d)	Print the dataframe who are Males and are Employed.

# Load the combined file
df = pd.read_csv("combined.csv")

# Filter: Males AND Employed
male_employed = df[(df["Gender"] == "Male") & (df["Employed"] == "yes")]

print(male_employed)
```

```python
#e)	Create a new Column ‘IQ_level’ which will have 3 values (Intelligent, Mediocre, weak).
#If student scored than 80 then Tag him as Intelligent, if student scored between 50-80, then Mediocre, else weak.

# Load the combined file
df = pd.read_csv("combined.csv")

# Create IQ_level column using conditions
df["IQ_level"] = df["Mark"].apply(
    lambda x: "Intelligent" if x > 80 else ("Mediocre" if x >= 50 else "Weak")
)

print(df.head(232))
```

```python
#f)	Count the number of males and females from each city.

# Load the combined file
df = pd.read_csv("combined.csv")

# Group by City and Gender, then count
gender_count = df.groupby(["City", "Gender"]).size().reset_index(name="Count")

print(gender_count)
```

```python
#g)	Print the top 5 Male scorers.

# Load the combined file
df = pd.read_csv("combined.csv")

# Filter only Male students and sort by Marks descending
top_male_scorers = df[df["Gender"] == "Male"].sort_values(by="Mark", ascending=False).head(5)

print(top_male_scorers[["Student_id", "Mark", "City", "Age", "Grade"]])
```

```python
#h)	Replace the Male value with M and Female value with F and export this dataframe to excel file in D: (D drive) and name the file as test.csv.

# Load the combined file
df = pd.read_csv("combined.csv")

df["Gender"] = df["Gender"].replace({"Male": "M", "Female": "F"})

df.to_csv("D:/test.csv", index=False)

print("File saved successfully at D:/test.csv")
```

```python
#k)	Get those Student_IDs which are present in Students table but not in Marks table.

# Load both CSVs
students_df = pd.read_csv("student.csv")

# Get IDs present in Students but not in Marks
missing_ids = students_df[~students_df["Student_id"].isin(marks_df["Student_id"])]

print("Student IDs present in Students table but not in Marks table:")
print(missing_ids["Student_id"].values)
```

```python
# Sample DataFrame
data = {'Name': ['A', 'B', None, 'D'],
        'Age': [25, None, 30, 40],
        'City': ['Delhi', 'Mumbai', None, 'Chennai']}
df = pd.DataFrame(data)

# Check for missing values
print(df.isnull())          # Boolean mask
print(df.isnull().sum())    # Count missing values per column
print(df.isnull().sum().sum())  # Total missing values


data = {'Course': ['Python', 'Pandas', 'SQL', 'Power BI'],
        'Status': ['Available', 'Seats are full!!', None, 'Available']}
df = pd.DataFrame(data)

# Replace "Seats are full!!" with a standard label
df['Status'] = df['Status'].replace("Seats are full!!", "Full")

# Identify missing values
print(df.isnull().sum())

# Fill missing values with "Unknown"
df['Status'] = df['Status'].fillna("Unknown")

print(df)
```
