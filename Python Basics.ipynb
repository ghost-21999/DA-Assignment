Q2. You’re a Data Scientist. You have assigned a task to perform. Create two lists l1 & l2, where l1 has values -> (1,2,3,4,5,6) 
and l2 has values -> (6,5,4,3,2,1). Compare the corresponding element values of both the lists 
and print “L1 element value is greater than L2 element value” if corresponding value from l1 is greater else 
print “L1 element value is less than L2 element value” if corresponding value from l1 is less than l2.


```python
# Create two lists
l1 = [1, 2, 3, 4, 5, 6]
l2 = [6, 5, 4, 3, 2, 1]

# Compare corresponding elements
for a, b in zip(l1, l2):
    if a > b:
        print(f"L1 element value {a} is greater than L2 element value {b}")
    elif a < b:
        print(f"L1 element value {a} is less than L2 element value {b}")
```

Q3. Write a Python user defined function that takes a list of numbers from user 
and returns a dictionary with the number as the key and its square as the value.


```python
# User defined function
def square_dictionary(numbers):
    """
    Function takes a list of numbers from user
    and returns a dictionary with the number as key
    and its square as the value.
    """
    result = {}
    for num in numbers:
        result[num] = num ** 2
    return result


# --- Main Program ---
# Taking list of numbers from user
user_input = input("Enter numbers separated by spaces: ")

# Converting input string into list of integers
num_list = [int(x) for x in user_input.split()]

# Calling the function
output = square_dictionary(num_list)

# Printing the resulting dictionary
print("Resulting dictionary:", output)
```

Q4. You work in a Company as a Analyst. The company officials want you to build a python based Calculator. 
Write code to create a user defined function named Calculator.
●	User will pass 3 parameters in Calculator function – int1, int2, task
●	The task string will be ‘add’, ‘sub’, ‘mul’, ‘div’. Perform Addition, Subtraction, Multiplication and Division on the two numbers(int1, int2) 
    respectively. Make sure the task parameter is case insensitive (meaning that even if user passes ‘aDd’ as 
    command it should perform addition on two numbers).
●	Define methods for addition, subtraction, multiplication and division in 
    Calculator function that take in two numbers and return another number after performing their respective operations.
Eg: Calculator(2,5,ADD) should return 7  Calculator(7,1,SuB) should return 6


```python
# User defined function named Calculator
def Calculator(int1, int2, task):
    """
    Function performs Addition, Subtraction, Multiplication, Division
    based on the task parameter. Task is case-insensitive.
    """

    # Define methods inside Calculator
    def add(a, b):
        return a + b

    def sub(a, b):
        return a - b

    def mul(a, b):
        return a * b

    def div(a, b):
        if b == 0:
            return "Error: Division by zero"
        return a / b

    # Convert task to lowercase for case-insensitivity
    task = task.lower()

    # Perform operation based on task
    if task == "add":
        return add(int1, int2)
    elif task == "sub":
        return sub(int1, int2)
    elif task == "mul":
        return mul(int1, int2)
    elif task == "div":
        return div(int1, int2)
    else:
        return "Invalid task provided"

# --- Sample Runs ---
print("Calculator(2, 5, 'ADD') =", Calculator(2, 5, 'ADD'))
print("Calculator(7, 1, 'SuB') =", Calculator(7, 1, 'SuB'))
print("Calculator(4, 3, 'Mul') =", Calculator(4, 3, 'Mul'))
print("Calculator(10, 2, 'div') =", Calculator(10, 2, 'div'))
```

Q5 Write a Function to check if the year number is a leap year or not. Print all Leap Years of 21st Century.


```python
# Function to check if a year is a leap year
def is_leap_year(year):
    """
    Returns True if the year is a leap year, else False.
    """
    if (year % 400 == 0) or (year % 100 != 0 and year % 4 == 0):
        return True
    else:
        return False


# --- Main Program ---
print("Leap Years of the 21st Century are:")

for year in range(2001, 2100):   # 21st century years
    if is_leap_year(year):
        print(year)
```

Q8 . Write a Function to take a list and return 2 lists:
-	1st list should contain all even indexed numbers.
-	2nd list should contain all odd indexed numbers.


```python
# Function to split list into even-indexed and odd-indexed elements
def split_even_odd(lst):
    """
    Takes a list and returns two lists:
    - First list contains all even indexed numbers
    - Second list contains all odd indexed numbers
    """
    even_indexed = []
    odd_indexed = []

    for i in range(len(lst)):
        if i % 2 == 0:   # Even index
            even_indexed.append(lst[i])
        else:            # Odd index
            odd_indexed.append(lst[i])

    return even_indexed, odd_indexed


# --- Main Program ---
# Example input list
numbers = [10, 20, 30, 40, 50, 60, 70]

# Call the function
even_list, odd_list = split_even_odd(numbers)

# Print results
print("Original list:", numbers)
print("Even indexed list:", even_list)
print("Odd indexed list:", odd_list)
```
